## 	베르누이 분포(Bernoulli distribution) 확률변수 X 는 0이나 1과 같은 두 가지 값만 가지게 되는 확률변수이다. X 값이 1이 될 확률을 p라고 하면, X=0일 확률은 1-p가 된다. X 가 확률 p인 베르누이 분포를 갖는다고 할 때, 평균과 분산을 구하여라.

```python
from sympy.stats import Bernoulli, E, density
from sympy import symbols

p=symbols('p',negative=False) # p를 음수가 아닌 기호로 정의

X=Bernoulli('X',p)

# 확률질량함수
k=symbols('k',integer=True)
density(X)(k)
```

```python
# E(X)는 X의 평균을 구해주는 파이썬 함수
E(X)
```

```python
# X**2의 평균
E(X**2)
```

```python
# X의 분산
V=E(X**2)-E(X)**2
V
```

```python
# X의 분산(내부함수이용)
from sympy.stats import variance
variance(X)
```

```python
_.simplify()
```

```python
# 이항분포에서 n=1인 경우가 베르누이 분포
from sympy.stats import Binomial

p=symbols('p',negative=False)
k=symbols('k',integer=True)
X=Binomial('X',1,p)

#확률질량함수
density(X)(k)
```

```python
# 평균
E(X)
```

```python
# 분산
variance(X).simplify()
```

## 이항분포 Xi (i=1,…,n)가 각각 서로 독립적인 베르누이 분포를 따른다고 할 때, X=X1+…+Xn이 따르는 분포를 이항분포라고 한다. 1회 시행시 성공확률을 p, n회 수행한다고 할 때, X는 B(n,p)를 따른다고 표기한다. X의 확률질량함수를 구하여라.

```python
from sympy.stats import Binomial,density

p=symbols('p',negative=False)
n,k=symbols('n k',integer=True)
X=Binomial('X',n,p) # X~B(n,p)인 이항분포 정의
```

```python
# X의 확률질량함수
density(X)(k)
```
